```python


import numpy as np
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
import warnings

# suppresses pandas warnings
warnings.filterwarnings(
    "ignore",
    message=r"datetime\.datetime\.utcnow\(\) is deprecated",
    category=DeprecationWarning
)

import pandas as pd
```

```python
df1 = pd.read_csv('grocerybasket.csv')
df1.head()
```

```python
df1.drop(columns=[df1.columns[0]],inplace=True)
market_basket = (
    pd.get_dummies(df1.stack())      # Flattens the df and creates one-hot columns for each item
    .groupby(level=0)               # Groups back by the original row index
    .max()                          # Keeps a 1 if the item was in that row, 0 otherwise
    .reindex(df1.index, fill_value=0) # Ensures rows that were completely empty are kept as 0
)
market_basket.head()
```
