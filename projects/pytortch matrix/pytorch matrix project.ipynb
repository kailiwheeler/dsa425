<h2>Pytorch Matrix Project</h2>

```python
import matplotlib.pyplot as plt
import numpy as np
import torch
```

read s, v, and u in as numpy arrays

```python
v = np.loadtxt('v.csv', delimiter=',', skiprows=0)
s_diag = np.loadtxt('s.csv', delimiter=',', skiprows=0)
u = np.loadtxt('u.csv', delimiter=',', skiprows=0)
s = np.diag(s_diag)
```

convert s, v, and u to pytorch tensors

```python
new_v = torch.from_numpy(v)
new_u = torch.from_numpy(u)
new_s = torch.from_numpy(s)
```

multiply the 3 tensors

```python
u_times_s = torch.mm(new_u,new_s)
multiplied_matrix = torch.mm(u_times_s,new_v)
print(multiplied_matrix)
```

```python
M = multiplied_matrix
plt.subplots(figsize=(3, 3)) # choose size of plot
plt.axis('off') # turn off axis numbering
plt.imshow(M, cmap=plt.cm.gray ) # plot the image
```

reconstruct compressed matrix, reduce rank

```python
k = 10
sliced_s = new_s[0:k,0:k]
sliced_u = new_u[:,0:k]
sliced_v = new_v[0:k,:]
u_times_s_sliced = torch.mm(sliced_u,sliced_s)
compressed_matrix = torch.mm(u_times_s_sliced,sliced_v)
```

```python
M = compressed_matrix
plt.subplots(figsize=(3, 3)) # choose size of plot
plt.axis('off') # turn off axis numbering
plt.imshow(M, cmap=plt.cm.gray ) # plot the image
```
